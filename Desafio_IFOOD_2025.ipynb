Nota-se que os nulos indicados acima podem significar que a pessoa não tem aquela determinada característica. Vamos exibi-las abaixo para verificar melhor esse ponto.

```python
# Selecionando apenas as colunas desejadas
colunas_selecionadas = ['marital_Divorced', 'marital_Married', 'marital_Single', 'marital_Together', 'marital_Widow',
                        'education_2n Cycle', 'education_Basic', 'education_Graduation', 'education_Master', 'education_PhD']
df_selecionado = df[colunas_selecionadas]

# Exibindo os valores das colunas selecionadas
df_selecionado
```

Observa-se que essas colunas só tem nulos ou um outro determinado valor. Certamente são variáveis categóricas nominais. Nesse caso o mais indicado é transforma-las em Booleanos onde 1 podemos dizer que fez e 0 que não fez (nulo).

```python
#selecionando colunas com nulos
colunas_com_nulos = ["marital_Divorced"
, "marital_Married"
, "marital_Single"
, "marital_Together"
, "marital_Widow"
, "education_2n Cycle"
, "education_Basic"
, "education_Graduation"
, "education_Master"
, "education_PhD"]
```

```python
#substituindo por booleano

for item in colunas_com_nulos:
  df["booleano"+str(item)] = np.where(df[item].isnull(), 0, 1) #se for nulo, substituimos por 0, caso contrário substituimos por 1
```

Para confirmar diferenças significativas entre os grupos, seria necessário um teste de hipótese.

# As pessoas gastam mais ou menos em nossa plataforma quando têm filhos? Veja nas colunas expenses e kids

```python
sns.boxplot(x='kids', y='expenses', data=df)
```

Vamos calcular a média de expenses por status de união

```python
#calculando a média de gastos
df.groupby("kids")["expenses"].mean()
```

```python
#calculando a mediana
df.groupby("kids")["expenses"].median()
```

Podemos ver que tanto a média quanto a mediana 0 e 1 filho é maior nessa amostra de dados, porém a média de 2 filhos é menor do que de 3 filhos, enquanto a mediana de 2 filhos é maior do que de 3 filhos. Ambos os grupos 2 e 3 tem médias bem maiores do que a mediana, indicando outliers superiores (o próprio gráfico já mostra isso). Grupo 1 também tem outliers superiores, mas essa diferença é menos acentuada. Pessoas com nenhum filho tem uma distribuição um pouco mais similar a uma normal (sem outliers, e mediana levemente centrada do meio do boxplot)
